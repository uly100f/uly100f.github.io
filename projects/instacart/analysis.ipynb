```python
proporcion_reorden = df_order_products.groupby('product_id')['reordered'].mean().reset_index()
```

```python
proporcion_reorden.columns = ['product_id', 'reorder_proportion']
```

```python
proporcion_reorden = proporcion_reorden.merge(
    df_products[['product_id', 'product_name']],
    on='product_id',
    how='left'
)

proporcion_reorden = proporcion_reorden[
    ['product_id', 'product_name', 'reorder_proportion']
]


proporcion_reorden = proporcion_reorden.sort_values(
    'reorder_proportion',
    ascending=False
)

print("Vista superior:")
print(proporcion_reorden.head(10))

print("\nVista inferior:")
print(proporcion_reorden.tail(10))
```

```python
df_order_products_con_user = df_order_products.merge(
    df_instacart_orders[['order_id', 'user_id']],
    on='order_id',
    how='left'
)
```

```python
cliente_reorden = (
    df_order_products_con_user
    .groupby('user_id')
    .agg(
        total_products=('reordered', 'count'),
        reordered_products=('reordered', 'sum')
    )
)

cliente_reorden['reorder_rate'] = (
    cliente_reorden['reordered_products'] / cliente_reorden['total_products']
)

cliente_reorden = cliente_reorden.sort_values('reorder_rate', ascending=False)

print("Vista superior:")
print(cliente_reorden.head(10))
```

```python
first_items = df_order_products[df_order_products['add_to_cart_order'] == 1]
```

```python
top_first_items = (
    first_items
    .groupby('product_id')
    .size()
    .reset_index(name='first_added_count')
)
top_first_items = top_first_items.merge(
    df_products[['product_id', 'product_name']],
    on='product_id'
)
```

```python
top_20 = top_first_items.sort_values(
    by='first_added_count',
    ascending=False
).head(20)

print(top_20)
```
